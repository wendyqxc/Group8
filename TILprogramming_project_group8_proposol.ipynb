# 1. Introduction

COVID-19 is the infectious disease that emerged at the end of 2019. Its consequences were visible in the Netherlands from January 2020 until May 2022. As a result, the government introduced measures such as keeping 1.5 meters distance, wearing face masks, encouraging people to work from home as much as possible, banning contact professions, and closing schools. These measures have led to significant improvements in technology for remote learning and working, which has resulted in changes in people's travel behaviour. 

This project investigates how the travel behaviour of Dutch people has changed after the pandemic compared to before the pandemic. The research focuses on people's travel behaviour between urban and rural areas. It looks at changes in trip frequency, distance, and purpose. Furthermore, it examines the role of remote work and digitalization. This question is important because transport systems and road capacity are planned years in advance. This research provides insights into which shifts caused by the pandemic persist across different population groups. As a result, the government gained a better understanding to optimize the Dutch infrastructure with respect to roads and railways.

# 2. Research Objective

**"How did the COVID-19 pandemic reshape travel behaviour in urban versus rural regions of the Netherlands?”**

The main objective of this research is to examine whether COVID-19 influenced the travel behaviour of Dutch citizens. To capture this potential impact, we will compare data from 2019 (pre-pandemic) with data from 2023 (post-pandemic). By contrasting these two periods, we aim to identify whether travel patterns have shifted as a result of COVID-19.

To refine the analysis, a distinction is made between two segments of the Dutch population: residents of urban areas and those in rural areas. This division is particularly relevant, as the spatial context strongly shapes mobility patterns. In urban areas, higher population density and closer proximity of facilities often lead to shorter and more frequent trips, while in rural areas longer distances between destinations typically result in different travel needs and behaviours. Comparing these contexts provides insights into how the pandemic may have reshaped travel behaviour across varying spatial environments.

This research does not distinguish between different modes of transport, but instead analyses overall travel behaviour. This approach is chosen because the available datasets only allow for reliable measurement of total mobility, rather than separate modal splits.
Sub objectives: 
1. Is there a difference in the average distance travelled per trip on each day of the week between urban and rural residents in 2019 and 2023?
2. Is there a difference in the motives for travel (work versus leisure) between urban and rural residents of the Netherlands in 2019 and 2023?
3. Is there a difference in the amount of transport used between students from rural and urban areas in the Netherlands between 2019 and 2023?

We include the sub-questions above because these perspectives capture important distinctions in travel behaviour that help answer the main research question. Travel behaviour often differs between weekdays and weekends, as weekdays are usually linked to work or school and weekends more to leisure. Looking at these patterns helps to see how routines may have changed between 2019 and 2023. Travel motives can further explain these changes, since the pandemic led to more working from home, which likely reduced commuting and shifted the balance between work- and leisure-related trips. Studying these aspects for urban and rural residents shows whether the impact of COVID-19 differed across contexts.

## 4. Results
The project's main research question focuses on differentiating travel behavior between citizens in rural and urban areas. A map has been created to illustrate this distinction, categorizing provinces into five levels of urbanization based on the number of inhabitants per square kilometer:

- Very strongly urban: ≥ 2 500 addresses per km² 
- Strongly urban: 1 500 – 2 500 addresses per km² 
- Moderately urban: 1 000 – 1 500 addresses per km² 
- Slightly urban: 500 – 1 000 addresses per km² 
- Not urban: < 500 addresses per km²

It seems the highest level of urbanization is located on the west side of the Netherlands, and the lowest level of urbanization is located on the (north) east side. The westside contains the Netherlands’ major cities (Amsterdam, Rotterdam, The Hague and Utrecht) and their suburbs, which is called the “Randstad” (“Edge City”). According to the CBS (2024) the strongest growth in population has been around these major cities. Possible explanations are more people are moving towards the suburbs, rather than from the suburbs. As well as, higher birth rates. However, the number of inhabitants on the outskirts of the Netherlands has declined.

### Subquestion 1
**Is there a difference in the average distance travelled per trip on each day of the week between urban and rural residents in 2019 and 2023?**

To answer subquestion 1, Data Set 1 have been used to create different line graphs. These visualizations are crucial for breaking down travel behaviour into its key components: time (day of the week), geography (urbanity level), and the pre- vs. post-pandemic periods (2019 vs. 2023). 

The first pair of line graphs (‘Average distance traveled per day in 2019’ and ‘Average distance traveled per day in 2023’) are the key figures in showing the change in travel behaviour. On the x-axis the different days of the week are placed, on the y-axis the average distance traveled per trip (km). So it directly plots the average travel distance per trip for the five distinct levels of urbanism. By separating these groups and presenting the data for 2019 and 2023, we can directly compare urban and rural behaviours and identify changes between the two years for each day.

**Rural vs. Urban difference:**
One of the key conclusions we can see from these graphs is that there is a clear consistent difference in the travel distance per trip based on urbanity. The COVID-19 pandemic did not have an impact on this, and there is no shift in this behaviour between 2019 and 2023. The purple line (‘Not Urban’) has the highest average travel distance per trip for every single day of the week (2019 and 2023). The blue line (‘Very Strongly Urban’) has the lowest average travel distance per trip (2019 and 2023). The other degrees or urbanism fall predictably in between. This shows a clear gradient from shortest trips in cities to longest trips in rural areas.  

**Weekly pattern:**
The graphs show a very consistent weekly pattern for both pre- and postpandemic. This pattern is also regardless of the level of urbanism. It can be seen that during the weekdays there is a very similar amount of kilometers traveled (Monday to Thursday) with a slight dip on Friday. Post pandemic this decline is a bit larger then pre pandemic, especially for the ‘Very Strongly Urban’ group (blue line). This is then followed by a large increase in average kilometers travelled for the weekend. With the largest increase on Sundays. The largest increase being for rural citizens pre pandemic, and post pandemic the largest increase is for the urban citizens.
